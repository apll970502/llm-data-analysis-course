### 실습 2. 분석할 제목 문자열 준비하기

Vectorizer에 전달할 제목을 문자열 형태로 준비합니다.

```python
import pandas as pd

# (참고) df_books 데이터프레임이 이전 단계에서 이미 로드되어 있다고 가정합니다.
# 예: df_books = pd.read_csv("book_bestseller_clean.csv")

# 1. '상품명' 컬럼의 텍스트 데이터 전처리
titles = (
    df_books["상품명"]
    .fillna("")        # 결측치(비어있는 값)를 빈 문자열("")로 채워 오류를 방지합니다.
    .astype(str)       # 모든 데이터를 안전하게 문자열(String) 타입으로 변환합니다.
    .str.strip()       # 문자열 앞뒤에 있는 불필요한 공백을 깔끔하게 제거합니다.
)

# 2. 전처리 후 빈 문자열("")이 되어버린 데이터를 제외하고, 인덱스를 처음부터 다시 맞춥니다.
titles = titles[titles != ""].reset_index(drop=True)

# 3. 최종적으로 분석에 사용할 제목의 개수를 출력하여 확인합니다.
print("사용할 제목 수:", len(titles))

# 4. 결과가 잘 담겼는지 상위 10개의 데이터를 확인합니다.
titles.head(10)
```
