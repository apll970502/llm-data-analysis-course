```python
products = [
    {"name": "키보드", "price":45000, "stock":12, "category":"전자기기"},
    {"name": "마우스", "price":45000, "stock":0, "category":"전자기기"},
    {"name": "모니터", "price":45000, "stock":4, "category":"전자기기"},
    {"name": "책상", "price":45000, "stock":0, "category":"가구"},
    {"name": "의자", "price":45000, "stock":7, "category":"가구"},
    {"name": "키보드", "price":45000, "stock":5, "category":"전자기기"},
    {"name": "마우스", "price":45000, "stock":2, "category":"전자기기"},
    {"name": "의자", "price":45000, "stock":3, "category":"가구"},
]

print(products)
```

```python
for_sales = []

found = False

for product in products:
    if product["stock"] >= 1:
        found = False
        for sale in for_sales:
            if sale["name"] == product["name"]:
                sale["stock"] += product["stock"]
                found = True
                break
        if not found:
            for_sales.append(product.copy())
            
for_sales    
```

### 실습 8. 문자열 앞뒤 공백 정리하기

상품명, 저자, 출판사, 분야는 문자열 데이터입니다.

문자열 앞뒤에 불필요한 공백이 있으면 같은 값도 서로 다른 데이터처럼 처리될 수 있습니다.

예를 들어 다음 두 값은 사람이 보기에는 같지만 문자열로는 다릅니다.

```python
# 1. 공백을 제거할 문자열 컬럼 이름들을 리스트로 묶어줍니다.
text_columns = ['상품명', '저자', '출판사', '분야']

# 2. for 반복문을 이용해 컬럼을 하나씩 꺼내면서 공백을 제거합니다.
for col in text_columns:
    # str.strip(): 문자열(str)의 양 끝에 있는 공백을 잘라냅니다(strip).
    df_books[col] = df_books[col].str.strip()

# 3. 처리가 끝난 후, 해당 컬럼들의 앞의 5개 값을 확인합니다.
df_books[text_columns].head()
```

### 🧹 문자열 공백 제거 결과 요약

* **처리 완료**: 지정한 4개 컬럼(`상품명`, `저자`, `출판사`, `분야`) 데이터의 양끝에 숨어있던 불필요한 공백(띄어쓰기)이 모두 성공적으로 제거되었습니다.
* **데이터 확인**: 상위 5개 행을 확인해 본 결과, 텍스트가 밀리거나 이상한 여백 없이 깔끔하게 정돈된 것을 볼 수 있습니다.
* **의미**: 겉보기엔 큰 차이가 없지만, 데이터 분석에서는 아주 중요한 작업입니다! 이제 컴퓨터가 단어를 정확하게 인식하게 되어, 앞으로 '소설' 분야의 책이 몇 권인지 세거나 특정 저자의 책을 검색할 때 띄어쓰기 때문에 누락되는 오류를 방지할 수 있습니다.

### 실습 9. 중복 데이터 확인하기

 

데이터에서 중복을 발견했다고 해서 바로 삭제하면 안 됩니다.

 

먼저 무엇을 기준으로 중복인지 생각해야 합니다.

 

예를 들어 같은 상품명이 여러 번 등장할 수 있습니다.

 

하지만 같은 제목이라도 다른 판본, 다른 상품, 개정판 등이 있을 수 있습니다.

 

따라서 이번 데이터에서는 상품명만 보고 중복 행을 바로 삭제하지 않습니다.

 

우선 상품을 구분하는 판매상품ID를 확인합니다.

```python
# 1. 전체 행(모든 컬럼의 값)이 완전히 똑같은 중복 데이터 개수 확인
total_duplicates = df_books.duplicated().sum()
print(f"전체 행 중복 개수: {total_duplicates}개")

# 2. '판매상품ID'만 기준으로 중복된 데이터 개수 확인
id_duplicates = df_books.duplicated(subset=['판매상품ID']).sum()
print(f"판매상품ID 중복 개수: {id_duplicates}개")

# 3. '상품명'만 기준으로 중복된 데이터 개수 확인
name_duplicates = df_books.duplicated(subset=['상품명']).sum()
print(f"상품명 중복 개수: {name_duplicates}개")
```
