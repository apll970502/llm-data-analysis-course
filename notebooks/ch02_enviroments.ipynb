1. 실행 환경 확인
먼저 현재 Notebook 커널과 작업 폴더를 확인합니다. VS Code에서 프로젝트의 .venv 커널을 선택한 뒤 실행하세요.

```python
from pathlib import Path
import sys

print('Python 실행 파일:', sys.executable)
print('현재 작업 폴더:', Path.cwd())
```

2. 패키지와 데이터 경로 설정
Notebook의 현재 작업 폴더가 프로젝트 루트인지 notebooks 폴더인지에 따라 상대 경로가 달라질 수 있습니다. 아래 코드는 두 경우를 모두 확인해 데이터 폴더를 선택합니다.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path
import sys

# ---- 프로젝트 루트 찾기 (노트북을 어느 폴더에서 열어도 동작한다) ----
# 현재 폴더에서 시작해 위로 한 단계씩 올라가며 data 폴더가 있는 곳을 찾는다.
PROJECT_ROOT = Path.cwd().resolve()

while not (PROJECT_ROOT / "data").is_dir():
    # 최상위(C:\)까지 올라왔는데도 못 찾으면 멈춘다. 이 검사가 없으면 무한 루프가 된다.
    if PROJECT_ROOT == PROJECT_ROOT.parent:
        raise FileNotFoundError("data 폴더를 찾지 못했습니다.")
    PROJECT_ROOT = PROJECT_ROOT.parent

# course_utils 를 import 할 수 있도록 프로젝트 루트를 검색 경로에 넣는다.
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

DATA_DIR = PROJECT_ROOT / "data" / "raw"

print("프로젝트 루트:", PROJECT_ROOT)
print("데이터 폴더:", DATA_DIR, "| 존재:", DATA_DIR.exists())

sns.set_theme(style="whitegrid")
```

3. 샘플 데이터 불러오기
customers.csv가 없다면 프로젝트 루트의 터미널에서 python scripts/generate_sample_data.py를 먼저 실행하세요.

```python
customers_path = DATA_DIR / 'customers.csv'

if not customers_path.exists():
    raise FileNotFoundError(
        f'{customers_path} 파일이 없습니다. '
        '프로젝트 루트에서 python scripts/generate_sample_data.py를 실행하세요.'
    )

customers = pd.read_csv(customers_path)
customers.head()
```

4. 기본 구조 확인
행과 열의 개수, 컬럼명, 데이터 타입을 확인합니다.

```python
print('데이터 크기:', customers.shape)
print('컬럼명:', customers.columns.tolist())
customers.info()
```
